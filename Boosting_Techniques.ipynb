#Theoretical questions
---

**1. What is Boosting in Machine Learning?**  
Boosting is an ensemble technique that combines multiple weak learners, typically decision trees, to create a strong learner. Each new model focuses on correcting the errors made by previous ones. For example, AdaBoost improves predictions by giving more weight to misclassified points.

---

**2. How does Boosting differ from Bagging?**  
Bagging builds multiple independent models in parallel and averages their outputs (e.g., Random Forest), while Boosting builds models sequentially, where each model tries to fix the errors of the previous one. Boosting usually leads to better accuracy but is more prone to overfitting.

---

**3. What is the key idea behind AdaBoost?**  
AdaBoost (Adaptive Boosting) aims to combine weak classifiers into a strong classifier. It assigns higher weights to misclassified instances so that subsequent models focus more on hard-to-classify examples. Over iterations, the model improves accuracy significantly.

---

**4. Explain the working of AdaBoost with an example.**  
Suppose a simple classifier incorrectly labels 30% of data points. AdaBoost increases the weights of these misclassified points. In the next iteration, the new classifier focuses more on these difficult cases. The final prediction is a weighted vote across all classifiers.

---

**5. What is Gradient Boosting, and how is it different from AdaBoost?**  
Gradient Boosting builds trees sequentially, where each new tree fits the residual errors (loss gradients) of the previous model. Unlike AdaBoost, which updates instance weights, Gradient Boosting minimizes a loss function directly through gradient descent, offering greater flexibility.

---

**6. What is the loss function in Gradient Boosting?**  
The loss function measures the difference between predicted and true values. Common loss functions include Mean Squared Error (for regression) and Log Loss (for classification). Gradient Boosting aims to minimize this loss at each stage by fitting new trees to the residuals.

---

**7. How does XGBoost improve over traditional Gradient Boosting?**  
XGBoost (Extreme Gradient Boosting) improves speed and performance with features like regularization, parallel processing, tree pruning, and handling missing values. For example, it uses second-order derivatives for faster optimization and better accuracy.

---

**8. What is the difference between XGBoost and CatBoost?**  
XGBoost requires manual preprocessing for categorical variables, while CatBoost handles categorical features automatically. CatBoost also reduces overfitting through techniques like Ordered Boosting, making it very efficient and accurate, especially with categorical-heavy datasets.

---

**9. What are some real-world applications of Boosting techniques?**  
Boosting is used in fraud detection, spam filtering, credit scoring, medical diagnosis, and recommendation systems. For example, financial companies use XGBoost models to predict loan defaults based on customer data.

---

**10. How does regularization help in XGBoost?**  
Regularization in XGBoost adds penalties to the complexity of trees (using L1 and L2 terms), preventing overfitting. It balances model bias and variance by discouraging overly complex trees that fit noise in the training data.

---

**11. What are some hyperparameters to tune in Gradient Boosting models?**  
Important hyperparameters include the learning rate, number of trees (n_estimators), maximum tree depth (max_depth), subsample ratio, and regularization parameters (lambda and alpha). Tuning them properly improves model performance significantly.

---

**12. What is the concept of Feature Importance in Boosting?**  
Feature Importance measures how valuable each feature is in making predictions. In Boosting, it is usually based on how often and how significantly a feature is used to split data across trees. It helps in feature selection and model interpretation.

---

**13. Why is CatBoost efficient for categorical data?**  
CatBoost efficiently handles categorical variables by converting them into numbers using techniques like Ordered Target Statistics and avoiding overfitting through Ordered Boosting. It eliminates the need for manual encoding (like one-hot), saving time and improving model performance.



```python
#14. Train an AdaBoost Classifier on a sample dataset and print model accuracy
from sklearn.ensemble import AdaBoostClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Generate a sample dataset
X, y = make_classification(n_samples=1000, n_features=20, random_state=42)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train AdaBoost Classifier
model = AdaBoostClassifier(n_estimators=50, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2f}")
```

```python
#15. Train an AdaBoost Regressor and evaluate performance using Mean Absolute Error (MAE)
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.datasets import make_regression

# Generate a regression dataset
X, y = make_regression(n_samples=1000, n_features=20, noise=0.1, random_state=42)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train AdaBoost Regressor
model = AdaBoostRegressor(n_estimators=50, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)

print(f"Mean Absolute Error: {mae:.2f}")
```

```python
#16. Train a Gradient Boosting Classifier on the Breast Cancer dataset and print feature importance
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import load_breast_cancer

# Load Breast Cancer dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Gradient Boosting Classifier
model = GradientBoostingClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Feature importance
importances = model.feature_importances_
feature_names = data.feature_names

# Print feature importances
for name, importance in zip(feature_names, importances):
    print(f"{name}: {importance:.4f}")
```

```python
#17. Train a Gradient Boosting Regressor and evaluate using R-Squared Score
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score

# Generate regression dataset
X, y = make_regression(n_samples=1000, n_features=20, noise=0.2, random_state=42)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Gradient Boosting Regressor
model = GradientBoostingRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)

print(f"R-Squared Score: {r2:.2f}")
```

```python
#18. Train an XGBoost Classifier on a dataset and compare accuracy with Gradient Boosting
from xgboost import XGBClassifier
from sklearn.ensemble import GradientBoostingClassifier

# Using Breast Cancer dataset again
data = load_breast_cancer()
X, y = data.data, data.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Gradient Boosting Classifier
gb_model = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb_model.fit(X_train, y_train)
gb_pred = gb_model.predict(X_test)
gb_acc = accuracy_score(y_test, gb_pred)

# XGBoost Classifier
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
xgb_model.fit(X_train, y_train)
xgb_pred = xgb_model.predict(X_test)
xgb_acc = accuracy_score(y_test, xgb_pred)

print(f"Gradient Boosting Accuracy: {gb_acc:.2f}")
print(f"XGBoost Accuracy: {xgb_acc:.2f}")
```

```python
#28. Train a CatBoost Classifier on an imbalanced dataset and compare performance with class weighting
from sklearn.datasets import make_classification
# Import the CatBoostClassifier
from catboost import CatBoostClassifier

# Create imbalanced data
X, y = make_classification(n_samples=1000, n_features=20, weights=[0.9, 0.1], random_state=42)
from sklearn.model_selection import train_test_split # Import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
from sklearn.metrics import f1_score # Import f1_score

# Without class weights
model_no_weight = CatBoostClassifier(verbose=0, random_state=42)
model_no_weight.fit(X_train, y_train)
f1_no_weight = f1_score(y_test, model_no_weight.predict(X_test))

# With class weights
model_weight = CatBoostClassifier(class_weights=[1,9], verbose=0, random_state=42)
model_weight.fit(X_train, y_train)
f1_weight = f1_score(y_test, model_weight.predict(X_test))

print(f"F1-Score without Class Weights: {f1_no_weight:.2f}")
print(f"F1-Score with Class Weights: {f1_weight:.2f}")
```

```python
#29.Train an AdaBoost Classifier and analyze the effect of different learning rates
import matplotlib.pyplot as plt # Import the pyplot module for plotting
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score

learning_rates = [0.01, 0.1, 0.5, 1.0]
accuracies = []


for lr in learning_rates:
    model = AdaBoostClassifier(learning_rate=lr, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracies.append(accuracy_score(y_test, y_pred))

# Plot
plt.plot(learning_rates, accuracies, marker='o')
plt.title("Effect of Learning Rates on Accuracy (AdaBoost)")
plt.xlabel("Learning Rate")
plt.ylabel("Accuracy")
plt.show()
```

```python
#30.Train an XGBoost Classifier for multi-class classification and evaluate using log-loss
from sklearn.datasets import make_classification
from sklearn.metrics import log_loss
from xgboost import XGBClassifier

# Multi-class dataset
X, y = make_classification(n_samples=1000, n_classes=3, n_informative=5, n_features=10, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# XGBoost multi-class
xgb_model = XGBClassifier(objective='multi:softprob', num_class=3, use_label_encoder=False, eval_metric='mlogloss', random_state=42)
xgb_model.fit(X_train, y_train)

# Predict and evaluate
y_pred_proba = xgb_model.predict_proba(X_test)
logloss = log_loss(y_test, y_pred_proba)

print(f"Log Loss (Multi-class XGBoost): {logloss:.2f}")
```
